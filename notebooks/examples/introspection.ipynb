```python
import pathlib

import networkx as nx

from vizopt import introspection
```

```python
src_dir = pathlib.Path().resolve().parents[1] / "src" / "vizopt"
```

```python
file_tree = introspection.build_file_tree(src_dir)
file_tree.edges
```

```python
introspection.plot_treemap(file_tree, padding=0.025)
```

```python
class_hierarchy = nx.DiGraph()
for node in file_tree.nodes:
    if file_tree.nodes[node]["is_dir"] or node.suffix != ".py":
        continue
    tree = introspection.parse_module_ast(src_dir / node)
    class_hierarchy = nx.compose(class_hierarchy, introspection.build_class_hierarchy(tree))

sorted(class_hierarchy.edges)
```

```python
import_graph = introspection.build_import_graph(src_dir)
sorted(import_graph.edges)
```
